## Ejemplo Plomero

```python
# --- SISTEMA DE GESTIÓN IA: PLOMERO ---

# Inicialización de las estructuras
plomero = []        # Lista para procesar datos individuales
losplomeros = {}    # Diccionario para almacenar a todos los plomeros

print("================================================")
print("BIENVENIDO AL ASISTENTE VIRTUAL 'PLOMERO-IA 2026'")
print("================================================")

# Ciclo infinito para mantener el menú interactivo
while True:
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Registrar nuevo plomero (Uso de Lista)")
    print("2. Recomendación de IA por especialidad")
    print("3. Ver base de datos (Diccionario de Listas)")
    print("4. Apagar sistema")

    opcion = input("\nSeleccione una opción (1-4): ")

    # OPCIÓN 1: REGISTRO
    if opcion == "1":
        print("\n--- REGISTRO DE TRABAJADOR ---")

        # Reiniciamos la lista para un nuevo registro
        plomero = []

        nombre = input("Nombre del plomero: ")
        especialidad = input("Especialidad (fuga/drenaje/sanitarios): ").lower()

        # Agregamos los datos a la lista
        plomero.append(nombre)
        plomero.append(especialidad)

        # Guardamos la lista en el diccionario
        losplomeros[nombre] = plomero

        print(f"IA: Datos de {nombre} guardados exitosamente.")

    # OPCIÓN 2: RECOMENDACIÓN IA
    elif opcion == "2":
        print("\n--- ASISTENTE DE DIAGNÓSTICO IA ---")

        if not losplomeros:
            print("IA: No hay personal registrado en el sistema.")
        else:
            busqueda = input("IA: ¿Qué especialidad requiere el cliente? ").lower()
            encontrado = False

            # Recorremos el diccionario
            for nombre_clave in losplomeros:

                # Accedemos a la lista guardada
                datos = losplomeros[nombre_clave]

                # Comparación usando condicional
                if datos[1] == busqueda:
                    print(f"\nIA RECOMIENDA: Para '{busqueda}', el especialista es {datos[0]} (Especialidad: {datos[1]}).")
                    encontrado = True

            if encontrado == False:
                print("IA: No se encontró un experto para esa área.")

    # OPCIÓN 3: MOSTRAR BASE DE DATOS
    elif opcion == "3":
        print("\n--- BASE DE DATOS ACTUAL ---")

        if not losplomeros:
            print("El diccionario está vacío.")
        else:
            for nombre, lista_datos in losplomeros.items():
                print(f"Plomero: {nombre} | Especialidad: {lista_datos[1]}")

    # OPCIÓN 4: SALIR
    elif opcion == "4":
        print("\nIA: Protocolo de cierre finalizado. ¡Hasta pronto!")
        break

    # ERROR
    else:
        print("\nError: Opción no válida. Intente de nuevo.")

print("\n--- PROGRAMA FINALIZADO ---")
print("Programa realizado por Iris Chavez NL = 11")
```
