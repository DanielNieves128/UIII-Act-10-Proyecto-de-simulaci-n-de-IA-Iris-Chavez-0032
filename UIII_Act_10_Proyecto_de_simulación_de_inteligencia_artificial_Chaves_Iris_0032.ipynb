## Ejemplo Plomero

```python
# --- SISTEMA DE GESTIÓN IA: PLOMERO-TECH V4.0 ---

# Inicialización de las estructuras según lo solicitado
plomero_actual = []  # Lista para procesar datos individuales
losplomeros = {}     # Diccionario para almacenar a todos los plomeros

print("================================================")
print("BIENVENIDO AL ASISTENTE VIRTUAL 'PLOMERO-IA 2026'")
print("================================================")

# Ciclo infinito para mantener el menú interactivo [5]
while True:
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Registrar nuevo plomero (Uso de Lista)")
    print("2. Recomendación de IA por especialidad")
    print("3. Ver base de datos (Diccionario de Listas)")
    print("4. Apagar sistema")

    opcion = input("\nSeleccione una opción (1-4): ")

    # OPCIÓN 1: REGISTRO USANDO UNA LISTA DENTRO DEL DICCIONARIO [2, 4]
    if opcion == "1":
        print("\n--- REGISTRO DE TRABAJADOR ---")
        # Reiniciamos la lista para un nuevo registro
        plomero_actual = []

        nombre = input("Nombre del plomero: ")
        especialidad = input("Especialidad (fuga/drenaje/sanitarios): ").lower()

        # Agregamos los datos a la lista temporal usando append [2]
        plomero_actual.append(nombre)
        plomero_actual.append(especialidad)

        # Guardamos la lista en el diccionario usando el nombre como clave [4]
        losplomeros[nombre] = plomero_actual
        print(f"IA: Datos de {nombre} guardados exitosamente.")

    # OPCIÓN 2: LÓGICA DE IA CON CICLO FOR [6, 7]
    elif opcion == "2":
        print("\n--- ASISTENTE DE DIAGNÓSTICO IA ---")
        # Verificamos si el diccionario tiene datos [8]
        if not losplomeros:
            print("IA: No hay personal registrado en el sistema.")
        else:
            busqueda = input("IA: ¿Qué especialidad requiere el cliente? ").lower()
            encontrado = False

            # Recorremos el diccionario para analizar los valores (que son listas) [9]
            for nombre_clave in losplomeros:
                # Accedemos a la lista guardada en esa clave
                datos = losplomeros[nombre_clave]

                # Comparamos la especialidad (índice 1 de la lista) con la búsqueda [2, 10]
                if datos[1] == busqueda: # Corrected index from 11 to 1
                    print(f"\nIA RECOMIENDA: Para '{busqueda}', el especialista es {datos[0]} (Especialidad: {datos[1]}).") # Adjusted print to be more readable
                    encontrado = True

            if encontrado == False:
                print("IA: No se encontró un experto para esa área.")

    # OPCIÓN 3: RECORRIDO COMPLETO DE DICCIONARIO [9]
    elif opcion == "3":
        print("\n--- BASE DE DATOS ACTUAL ---")
        if not losplomeros:
            print("El diccionario está vacío.")
        else:
            # Iteramos sobre los ítems del diccionario [7]
            for nombre, lista_datos in losplomeros.items():
                print(f"Plomero: {nombre} | Especialidad: {lista_datos[1]}") # Corrected index from 11 to 1

    # OPCIÓN 4: FINALIZAR PROGRAMA [5]
    elif opcion == "4":
        print("\nIA: Protocolo de cierre finalizado. ¡Hasta pronto!")
        break # Rompe el ciclo while True

    else:
        print("\nError: Opción no válida. Intente de nuevo.")

print("\n--- PROGRAMA FINALIZADO ---")
print("Programa realizado por Iris Chavez NL = 11")
```
